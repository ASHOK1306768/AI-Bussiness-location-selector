## 2. Discussion: Why MCDA is the primary decision backbone
Supervised regression identifies correlations (e.g. competitor agglomeration and parking), but ground-truth business performance data is scarce in emerging markets. Multi-Criteria Decision Analysis (MCDA) prevents overfitting and guarantees transparent, explainable recommendations.
