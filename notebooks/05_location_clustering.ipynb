# 05 - Unsupervised Location Clustering (K-Means)
**AI-Based Business Location Selector Project**

## 1. Objective
Cluster commercial zones using standardized features, evaluate Elbow & Silhouette curves, and profile commercial archetypes.

```
import os
import sys
sys.path.insert(0, '..')
import pandas as pd
import matplotlib.pyplot as plt
from src.preprocessing import LocationDataPreprocessor
from src.clustering import LocationClusterer

prep = LocationDataPreprocessor(data_dir='../data/processed')
df = prep.get_consolidated_dataset(save_output=False)
clusterer = LocationClusterer(n_clusters=4)
eval_res = clusterer.evaluate_cluster_range(df, k_range=range(2, 7))

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.plot(eval_res['k'], eval_res['inertia'], 'bo-')
plt.title('Elbow Method (Inertia)')
plt.xlabel('k (Clusters)')

plt.subplot(1, 2, 2)
plt.plot(eval_res['k'], eval_res['silhouette'], 'ro-')
plt.title('Silhouette Score Analysis')
plt.xlabel('k (Clusters)')
plt.show()
```

## 2. Fit K-Means and Examine Identified Profiles

```
clustered_df = clusterer.fit_predict(df)
for cid, name in clusterer.cluster_profiles.items():
    print(f'{name}: {(clustered_df["cluster_id"] == cid).sum()} locations')
```
