## 1. Objective
Explore distributions of rents, footfall, accessibility, and competitor density across pilot cities.

```
import os
import sys
sys.path.insert(0, '..')
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.preprocessing import LocationDataPreprocessor

prep = LocationDataPreprocessor(data_dir='../data/processed')
df = prep.get_consolidated_dataset(save_output=False)
df.describe()
```

## 2. Rental Comparison by City
Examine variation in commercial rents between Bhopal, Indore, and Pune.

```
plt.figure(figsize=(8, 4.5))
sns.boxplot(data=df, x='city', y='avg_monthly_rent', palette='Blues_d')
plt.title('Commercial Rent Distribution by City')
plt.ylabel('Monthly Rent (INR)')
plt.show()
```

## 3. Correlation Matrix
Analyze collinearity between demographic catchment, footfall, and rent.

```
plt.figure(figsize=(8, 6))
num_cols = ['population', 'competitor_density', 'footfall_estimate', 'accessibility_score', 'parking_score', 'avg_monthly_rent']
sns.heatmap(df[num_cols].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()
```
